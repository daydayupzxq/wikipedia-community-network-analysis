```python
import os
import pandas as pd

# 定义文件夹路径
folder_path = 'D:/Biyesheji1/stated_userdata'

# 初始化一个空的DataFrame
combined_df = pd.DataFrame()

# 遍历文件夹中的所有CSV文件
for file_name in os.listdir(folder_path):
    if file_name.endswith('.csv'):
        # 读取CSV文件
        file_path = os.path.join(folder_path, file_name)
        temp_df = pd.read_csv(file_path)
        
        # 添加文件名列
        temp_index = file_name.find('.csv')
        name = file_name[:temp_index]
    
        temp_df['File_Name'] =name
        if name not in list(df['File_Name']):
            print(name)
        
        # 将当前CSV文件的数据追加到combined_df中
        combined_df = pd.concat([combined_df,temp_df], ignore_index=True)
```

```python
import csv

def read_first_line_of_csv_files(folder_path):
    temp_list=[]
    # 获取文件夹下所有文件的列表
    file_list = os.listdir(folder_path)
    
    # 遍历文件列表
    for file_name in file_list:
        # 确保文件是CSV文件
        if file_name.endswith('.csv'):
            file_path = os.path.join(folder_path, file_name)
            # 打开CSV文件，指定编码方式为utf-8
            with open(file_path, 'r', newline='', encoding='utf-8') as csv_file:
                csv_reader = csv.reader(csv_file)
                # 读取第一行
                first_row = next(csv_reader)
                print([first_row[0][len('https://en.wikipedia.org/wiki/'):],first_row[3]])
                temp_list.append([first_row[0][len('https://en.wikipedia.org/wiki/'):],first_row[3]])
                
    return temp_list

# 调用函数并传入文件夹路径
folder_path = 'D:/Biyesheji1/wangyesuiji1000'
temp_list=read_first_line_of_csv_files(folder_path)
```

```python
def str_with_comma_to_int(str_num):
    # 去除逗号
    num_without_comma = str_num.replace(',', '')
    # 转换为整数
    num = int(num_without_comma)
    return num

# 测试示例
string_with_comma = "1,000,000"
result = str_with_comma_to_int(string_with_comma)
print(result)  # 输出结果为 1000000
```

```python
def has_number(input_string):
    if 'empty' not in input_string:
        for char in input_string:
            if char.isdigit():
                return True
    else:
        return False
```

```python
import pandas as pd


def standardize_column(group):
    # 计算分组的均值和标准差
    mean_val = group.mean()
    std_val = group.std()
    # 对分组内的值进行标准化处理
    standardized_values = (group - mean_val) / std_val
    return standardized_values

# 按照"grouping_column"进行分组，并对"column_to_normalize"进行标准化
combined_df['nnew_length'] = combined_df.groupby('File_Name')['new_length'].transform(standardize_column)

combined_df
```
