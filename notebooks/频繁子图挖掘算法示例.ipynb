```python
import pickle
def save_variable(v,filename):
    f=open(filename,'wb')
    pickle.dump(v,f)
    f.close()
    return filename
 
def load_variavle(filename):
    f=open(filename,'rb')
    r=pickle.load(f)
    f.close()
    return r

filename = save_variable(gs,'示例文件.txt')
results = load_variavle('示例文件.txt')
```

```python
import networkx as nx
import matplotlib.pyplot as plt

def turnto_graph(g):
    gnx = nx.Graph() if g.is_undirected else nx.DiGraph()
    vlbs = {vid: v.vlb for vid, v in g.vertices.items()}
    elbs = {}
    for vid, v in g.vertices.items():
        gnx.add_node(vid, label=v.vlb)
    for vid, v in g.vertices.items():
        for to, e in v.edges.items():
            if (not g.is_undirected) or vid < to:
                gnx.add_edge(vid, to, label=e.elb)
                elbs[(vid, to)] = e.elb
    fsize = (min(16, 1 * len(g.vertices)),
             min(16, 1 * len(g.vertices)))
    plt.figure(3, figsize=fsize)
    pos = nx.spectral_layout(gnx)
    nx.draw_networkx(gnx, pos, arrows=True, with_labels=True, labels=vlbs)
    nx.draw_networkx_edge_labels(gnx, pos, edge_labels=elbs)
    plt.show()
    
    return gnx
```
