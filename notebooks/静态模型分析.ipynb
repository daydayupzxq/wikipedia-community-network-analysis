### 含有多类角色的用户

```python
# 按用户名分组并统计每个用户拥有的唯一角色数量
grouped = df.groupby('username')['result'].nunique()

# 筛选出含有多类角色的用户名
multi_role_users = grouped[grouped > 1].index.tolist()

print("含有多类角色的用户名：", multi_role_users)
```

```python
len(multi_role_users)
```

```python
temp_data1=[]
temp_data2=[]

# 筛选出拥有多种角色的用户的数据
multi_role_df = df[df['username'].isin(multi_role_users)]

# 将每个用户的角色去重，创建角色组合
user_roles = multi_role_df.groupby('username')['result'].apply(lambda x: tuple(sorted(set(x))))

# 统计角色组合的频率
role_combinations = user_roles.value_counts()

# 输出角色组合出现的频率
print("多种角色组合的数量统计：")
for combination, count in role_combinations.items():
    roles_str = ', '.join(combination)
    temp_data1.append(roles_str)
    temp_data2.append(count)
    print(f"角色组合 {roles_str} 出现的数量为 {count}")
```

```python
temp_data1
```

```python
import matplotlib.pyplot as plt


# 设置不同柱子的颜色
colors = ['skyblue', 'orange', 'green', 'red', 'purple', 'yellow']

# 绘制柱状图
plt.figure(figsize=(10, 6))
bars = plt.bar(range(len(temp_data1)), temp_data2, color=colors)

# 在柱状图上显示数值
for i, count in enumerate(temp_data2):
    plt.text(i, count + 0.1, str(count), ha='center', va='bottom')

# 将图例放置在右上方
legend_labels = [str(i) for i in temp_data1]
legend_patches = [plt.Rectangle((0, 0), 1, 1, color=color) for color in colors]
plt.legend(legend_patches, legend_labels, loc='upper right', bbox_to_anchor=(1, 1), title='用户类型列表', frameon=False)

plt.ylabel('数量')
plt.title('重复用户列表数量分布')
plt.xticks([])  # 不显示 x 轴标注

# 显示图形
plt.tight_layout()
plt.savefig('重复用户列表数量分布.png')
plt.show()
```

```python
import matplotlib.pyplot as plt

# 统计不同用户类型的分布状况
user_type_distribution = df['result'].value_counts()

# 创建一个柱状图
plt.figure(figsize=(10, 6))
user_type_distribution.plot(kind='bar')

# 添加柱状图标签
for i, count in enumerate(user_type_distribution):
    plt.text(i, count + 0.1, str(count), ha='center', va='bottom')

plt.title('不同用户类型的分布')
plt.xlabel('用户类型')
plt.ylabel('计数')
plt.xticks(rotation=45)  # 旋转x轴标签，使其更易读
plt.tight_layout()  # 调整布局，防止标签被截断
plt.savefig('不同用户类型的分布图.png')
plt.show()
```

## 统计各词条下用户类型分布状况

```python
df.head()
```

```python
from collections import Counter

temp_list=[]
# 创建一个空的 Counter 对象
user_types_list_counter = Counter()

# 遍历每个文件名
for file_name, group in df.groupby('File_Name'):
    # 获取该文件中存在的用户类型列表
    existing_user_types = group['result'].unique()
    # 将用户类型列表转换为列表，以便进行统计
    user_types_list = sorted(existing_user_types.tolist())
    # 将列表转换为字符串，以便作为 Counter 的 key
    user_types_str = str(user_types_list)
    # 更新 Counter 对象
    user_types_list_counter.update([user_types_str])

# 输出统计结果
print("用户类型列表数量统计：")
for user_types_list, count in user_types_list_counter.items():
    # 将字符串转换回列表
    user_types_list = eval(user_types_list)
    temp_list.append(user_types_list)
    print(f"用户类型列表 {user_types_list} 的数量：{count}")
```

```python
import matplotlib.pyplot as plt

# 定义用户类型列表和对应的数量
counts = [966, 18, 7, 8, 1, 1,1]

# 设置不同柱子的颜色
colors = ['skyblue', 'orange', 'green', 'red', 'purple', 'yellow']

# 绘制柱状图
plt.figure(figsize=(6, 6))
bars = plt.bar(range(len(temp_list)), counts, color=colors)

# 在柱状图下方添加图例，并水平排列
legend_labels = [str(user_types_list) for user_types_list in temp_list]
legend_patches = [plt.Rectangle((0,0),1,1, color=color) for color in colors]
plt.legend(legend_patches, legend_labels, loc='upper center', bbox_to_anchor=(0.5, -0.1), ncol=1, title='用户类型列表', frameon=False)

plt.ylabel('数量')
plt.title('用户类型列表数量分布')
plt.xticks([])  # 不显示 x 轴标注

# 在柱状图上显示数值
for i, count in enumerate(counts):
    plt.text(i, count + 0.1, str(count), ha='center', va='bottom')

# 显示图形
plt.tight_layout()
plt.savefig('用户类型列表数量分布.png')
plt.show()
```

## 用户数量演化趋势

```python
import pandas as pd

# 读取数据到DataFrame
data =df[['username','topic','File_Name']]

# 按照文件名分组，统计每个文件名下的唯一用户数量
file_user_count = data.groupby('File_Name')['username'].nunique()

# 按照话题类型分组，找出每个话题下用户数量最多的文件名
topic_max_users = data.groupby('topic').apply(lambda x: x.groupby('File_Name')['username'].nunique().idxmax())

topic_max_users=topic_max_users.reset_index()
topic_max_users
```

```python
list1=list(topic_max_users[0])
list2=list(topic_max_users['topic'])
```

```python
df
```

```python
import matplotlib.pyplot as plt
import pandas as pd

list1=list(topic_max_users[0])
list2=list(topic_max_users['topic'])

# 创建一个三行四列的子图画布
fig = plt.figure(figsize=(12, 9))

# 遍历文件名列表并在每个子图中执行代码循环
for i in range(len(list2)):
    # 添加子图
    ax = fig.add_subplot(4,3, i+1)

    temp_data=df[df['topic']==(list2[i])]
    temp_data=temp_data[['Time','result']]
    
    # 将 '时间' 列转换为 datetime 类型
    temp_data['Time'] = pd.to_datetime(temp_data['Time'])

    # 按时间和用户类型排序
    temp_data.sort_values(by=['Time', 'result'], inplace=True)

    # 创建一个新的数据框来保存累积的用户数量
    cumulative_df = temp_data.groupby(['Time', 'result']).size().unstack(fill_value=0).cumsum()

    # 绘制折线图
    cumulative_df.plot(ax=ax)
    ax.set_xlabel('时间')
    ax.set_ylabel('累积用户数量')
    ax.set_title(f'{list2[i]}用户数量演化')
    ax.legend(title='用户类型', loc='upper left',bbox_to_anchor=(1.15, 1))
    ax.grid(True)

# 调整子图之间的间距
plt.tight_layout()

plt.savefig('用户数量演化趋势图.png')
# 显示图形
plt.show()
```
