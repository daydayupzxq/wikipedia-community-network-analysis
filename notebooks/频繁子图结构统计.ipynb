```python
import pickle
import networkx as nx
import matplotlib.pyplot as plt

# 读取存储的结构计数文件
structure_counts_file_path = 'D:/Biyesheji1/sum_structure_counts.pkl'
with open(structure_counts_file_path, 'rb') as f:
    structure_counts = pickle.load(f)


max_key = max(structure_counts, key=structure_counts.get)
max_key
```

```python
import pandas as pd

temp_data=pd.DataFrame(list(structure_counts.values()))
temp_data=temp_data.sort_values(by=0,ascending=False)
temp_data
```

```python
# 按值排序
sorted_keys = sorted(structure_counts, key=structure_counts.get, reverse=True)

# 输出排序后的键
for key in sorted_keys:
    print(key)
    
    # 定义一个包含节点名称修改映射的字典
    mapping = {0:"中坚支持者", 1:"中介成员", 2:"最近参与成员",3:'创新突破者',4.:'领导者'}

    # 使用字典修改节点名称
    H = nx.relabel_nodes(key, mapping)
    
    pos = nx.spring_layout(H)  # 选择一个布局算法，这里选择 Spring layout
    nx.draw(H, pos, with_labels=True, node_size=700, node_color="skyblue", font_size=12, font_weight="bold")
    plt.show()
```
