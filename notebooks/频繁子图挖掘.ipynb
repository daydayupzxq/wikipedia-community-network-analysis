```python
from collections import Counter

temp_list=[]
# 创建一个空的 Counter 对象
user_types_list_counter = Counter()

# 遍历每个文件名
for file_name, group in df.groupby('File_Name'):
    # 获取该文件中存在的用户类型列表
    existing_user_types = group['result'].unique()
    # 将用户类型列表转换为列表，以便进行统计
    user_types_list = sorted(existing_user_types.tolist())
    # 将列表转换为字符串，以便作为 Counter 的 key
    user_types_str = str(user_types_list)
    # 更新 Counter 对象
    user_types_list_counter.update([user_types_str])

# 输出统计结果
print("用户类型列表数量统计：")
for user_types_list, count in user_types_list_counter.items():
    # 将字符串转换回列表
    user_types_list = eval(user_types_list)
    temp_list.append(user_types_list)
    print(f"用户类型列表 {user_types_list} 的数量：{count}")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

def merge_nodes_by_type(G):
    """
    将图中相同类型的节点及其连接合并为超节点
    Args:
    - G: NetworkX 图对象

    Returns:
    - merged_G: 合并节点后的图对象
    """
    node_types = nx.get_node_attributes(G, 'category')  # 获取节点类型属性
    merged_G = nx.DiGraph()  # 创建一个有向图对象

    # 创建超节点
    super_nodes = {}
    for node, node_type in node_types.items():
        if node_type not in super_nodes:
            super_nodes[node_type] = [node]
        else:
            super_nodes[node_type].append(node)

    # 添加超节点及连接
    for super_node, nodes in super_nodes.items():
        merged_G.add_node(super_node, type='super_node')  # 添加超节点
        for node in nodes:
            for neighbor in G.neighbors(node):
                neighbor_type = node_types[neighbor]
                if neighbor_type == super_node:  # 如果邻居节点类型与超节点相同
                    continue
                if not merged_G.has_node(neighbor_type):  # 添加邻居节点类型
                    merged_G.add_node(neighbor_type, type='node_type')
                merged_G.add_edge(super_node, neighbor_type)  # 添加连接

    return merged_G

merged_G = merge_nodes_by_type(new_G)

# 绘制新的图对象
plt.figure(figsize=(8, 6))

# 定义节点颜色
node_color = ['blue' if merged_G.nodes[node]['type'] == 'super_node' else 'red' for node in merged_G.nodes()]

nx.draw(merged_G, with_labels=True, node_color=node_color, node_size=1000, font_size=12, font_weight='bold', arrows=True)

plt.title('Merged Directed Graph with Super Nodes')
plt.show()
```

```python
from collections import Counter

# 找到某两种类型节点间的最短路径并统计路径类型
def find_shortest_path_between_categories(G, category_1, category_2):
    shortest_paths = []
    for node_1 in nx.nodes(G):
        if G.nodes[node_1]['category'] == category_1:
            for node_2 in nx.nodes(G):
                if G.nodes[node_2]['category'] == category_2:
                    try:
                        shortest_path = nx.shortest_path(G, node_1, node_2)
                        shortest_path_categories = tuple(G.nodes[node]['category'] for node in shortest_path)
                        shortest_paths.append(shortest_path_categories)
                    except nx.NetworkXNoPath:
                        pass
    return shortest_paths

# 找到两种类型的节点彼此到达的最短路径并统计路径类型

category=['中介成员', '中坚支持者', '创新突破者', '最近参与成员', '领导者']
category_1 = category[1]  # 第一种节点类型
category_2 = category[3]  # 第二种节点类型

shortest_paths = find_shortest_path_between_categories(new_G, category_1, category_2)
```

```python
def simplify_and_filter_paths(path_types, category_2):
    simplified_paths = []
    for path in path_types:
        simplified_path = []
        for node_category in path:
            simplified_path.append(node_category)
            if node_category == category_2:
                break  # 当找到 category_2 后停止添加后续节点
        simplified_paths.append(tuple(simplified_path))
    return simplified_paths

def simplify_path_types(path_types):
    simplified_paths = []
    for path in path_types:
        simplified_path = [path[0]]
        for i in range(1, len(path)):
            if path[i] != path[i-1]:
                simplified_path.append(path[i])
        simplified_paths.append(tuple(simplified_path))
    return simplified_paths

# 简化路径类型
simplified_paths = simplify_path_types(shortest_paths)

# 过滤路径类型
filtered_paths = simplify_and_filter_paths(simplified_paths, category_2)

# 统计简化后的路径类型次数
filtered_path_type_counts = Counter(filtered_path_type for filtered_path_type in filtered_paths)

# 创建DataFrame存储简化后的路径类型和对应数量
df_filtered = pd.DataFrame(list(filtered_path_type_counts.items()), columns=['Path', 'Count'])

# 按数量排序
df_filtered = df_filtered.sort_values(by='Count', ascending=False)

# 重置索引
df_filtered = df_filtered.reset_index(drop=True)

df_filtered['Sour_Targ']=str([category_1, category_2])

df_filtered
```

```python
import networkx as nx


# 统计不同类型节点邻接节点的类型
type_neighbors = {}
for node in new_G.nodes(data=True):
    node_type = node[1]['category']
    neighbors = list(new_G.successors(node[0]))
    for neighbor in neighbors:
        neighbor_type = new_G.nodes[neighbor]['category']
        if node_type not in type_neighbors:
            type_neighbors[node_type] = {neighbor_type: 1}
        else:
            if neighbor_type not in type_neighbors[node_type]:
                type_neighbors[node_type][neighbor_type] = 1
            else:
                type_neighbors[node_type][neighbor_type] += 1


Temp_list1=[]
Temp_list2=[]
for node_type, neighbors_info in type_neighbors.items():
    print(f"Type: {node_type}")
    temp_list1=[]
    temp_list2=[]
    for neighbor_type, count in neighbors_info.items():
        print(f"  Neighbor Type: {neighbor_type}, Count: {count}")
        if not pd.Series(neighbor_type).isna().all():
            temp_list1.append(neighbor_type)
            temp_list2.append(count)
        
    Temp_list1.append(temp_list1)
    Temp_list2.append(temp_list2)
```
